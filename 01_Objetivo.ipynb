# 1 - Objetivo

## MVP - Engenharia de Dados | Performance de estudantes

Construir um pipeline de dados em arquitetura Medalhão — **Bronze, Silver e Gold** — para organizar, padronizar, validar e disponibilizar um conjunto de dados que consiste nas notas obtidas por alunos em diversas disciplinas, avaliando o desempenho acadêmico.

### Fonte
- Arquivo: `StudentsPerformance.xls`
- Link origem: (https://www.kaggle.com/datasets/spscientist/students-performance-in-exams?resource=download&select=StudentsPerformance.csv)
- Localização: **Databricks Workspace**, na pasta `MVP-Engenharia de Dados-MFarias`
- Granularidade: **1 registro = 1 estudante avaliado**

### Escopo
1. Preparação e documentação da fonte.
2. Disponibilização/validação da fonte no Workspace.
3. Ingestão preservando a estrutura original na Bronze.
4. Padronização, tipagem e regras de qualidade na Silver.
5. Organização dos dados analíticos na Gold.
6. Resposta às perguntas de negócio por SQL.


## Problema: Compreender o desempenho acadêmico de alunos em relação ao cotidiano, alimentação e conviencia familiar.

1. Qual é o desempenho médio dos estudantes nas avaliações de matemática, leitura e escrita?
2. Estudantes que realizaram o curso de preparação apresentam desempenho médio diferente dos estudantes que não realizaram?
3. Existe diferença no desempenho médio entre estudantes com almoço padrão e estudantes com almoço gratuito/reduzido?
4. Como o desempenho médio varia de acordo com o nível de escolaridade dos pais/responsáveis?
5. Como o desempenho médio varia entre os grupos de gênero presentes na base?
6. Quais combinações de curso de preparação, tipo de alimentação e escolaridade parental apresentam diferentes níveis de desempenho médio?
