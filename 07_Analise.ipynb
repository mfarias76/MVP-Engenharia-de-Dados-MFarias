# 7 - Análise de Dados

**Objetivo:** responder por meio de consultas SQL sobre a camada Gold, ao problema e perguntas definidas na etapa de objetivo.

**Fonte analítica:** `main.students_performance.students_performance_gold`

> **Nota:** as interpretações descrevem as associações observadas na base e não evidenciam causalidade.


## Pergunta 1 - Desempenho geral

**Qual é o desempenho médio dos estudantes nas avaliações de matemática, leitura e escrita?**

### Consulta SQL
A consulta retorna o total de estudantes e as médias de matemática, leitura, escrita e desempenho geral.

```python
%sql
SELECT COUNT(*) AS total_students, ROUND(AVG(math_score), 2) AS avg_math_score, ROUND(AVG(reading_score), 2) AS avg_reading_score, ROUND(AVG(writing_score), 2) AS avg_writing_score, ROUND(AVG(average_score), 2) AS avg_general_score FROM main.students_performance.students_performance_gold;
```

## Pergunta 2 - Curso de preparação

**Estudantes que realizaram o curso de preparação apresentam desempenho médio diferente dos estudantes que não realizaram?**

### Consulta SQL
A consulta retorna uma linha para cada situação: curso preparatório realizado (`completed`) e não realizado (`none`), com médias por disciplina e média geral.

```python
%sql
SELECT test_preparation_course, COUNT(*) AS total_students, ROUND(AVG(math_score), 2) AS avg_math_score, ROUND(AVG(reading_score), 2) AS avg_reading_score, ROUND(AVG(writing_score), 2) AS avg_writing_score, ROUND(AVG(average_score), 2) AS avg_general_score FROM main.students_performance.students_performance_gold GROUP BY test_preparation_course ORDER BY test_preparation_course;
```

## Pergunta 3 - Tipo de alimentação

**Existe diferença no desempenho médio entre estudantes com almoço padrão e estudantes com almoço gratuito/reduzido?**

### Consulta SQL
A consulta retorna uma linha para cada categoria de almoço (`lunch`), ou regular (`standard`) ou gratuito/reduzido (`free/reduced`).

```python
%sql
SELECT lunch, COUNT(*) AS total_students, ROUND(AVG(math_score), 2) AS avg_math_score, ROUND(AVG(reading_score), 2) AS avg_reading_score, ROUND(AVG(writing_score), 2) AS avg_writing_score, ROUND(AVG(average_score), 2) AS avg_general_score FROM main.students_performance.students_performance_gold GROUP BY lunch ORDER BY lunch;
```

## Pergunta 4 - Escolaridade dos pais/responsáveis

**Como o desempenho médio varia de acordo com o nível de escolaridade dos pais/responsáveis?**

### Consulta SQL
A consulta apresenta uma linha por nível de escolaridade, ordenada pelo desempenho geral médio.

```python
%sql
SELECT parental_level_of_education, COUNT(*) AS total_students, ROUND(AVG(math_score), 2) AS avg_math_score, ROUND(AVG(reading_score), 2) AS avg_reading_score, ROUND(AVG(writing_score), 2) AS avg_writing_score, ROUND(AVG(average_score), 2) AS avg_general_score FROM main.students_performance.students_performance_gold GROUP BY parental_level_of_education ORDER BY avg_general_score DESC;
```

## Pergunta 5 - Gênero

**Como o desempenho médio varia entre os grupos de gênero presentes na base?**

### Consulta SQL
A consulta retorna uma linha para cada categoria de gênero (`gender`), com médias das três disciplinas e média geral.

```python
%sql
SELECT gender, COUNT(*) AS total_students, ROUND(AVG(math_score), 2) AS avg_math_score, ROUND(AVG(reading_score), 2) AS avg_reading_score, ROUND(AVG(writing_score), 2) AS avg_writing_score, ROUND(AVG(average_score), 2) AS avg_general_score FROM main.students_performance.students_performance_gold GROUP BY gender ORDER BY gender;
```

## Pergunta 6 - Combinação de fatores

**Quais combinações de curso de preparação, tipo de alimentação e escolaridade parental apresentam diferentes níveis de desempenho médio?**

### Consulta SQL
A consulta retorna as combinações com pelo menos 10 estudantes, ordenadas pelo desempenho geral médio.

```python
%sql
SELECT test_preparation_course, lunch, parental_level_of_education, COUNT(*) AS total_students, ROUND(AVG(math_score), 2) AS avg_math_score, ROUND(AVG(reading_score), 2) AS avg_reading_score, ROUND(AVG(writing_score), 2) AS avg_writing_score, ROUND(AVG(average_score), 2) AS avg_general_score FROM main.students_performance.students_performance_gold GROUP BY test_preparation_course, lunch, parental_level_of_education HAVING COUNT(*) >= 10 ORDER BY avg_general_score DESC;
```

As consultas SQL sobre a camada Gold, apresentam informações que respondem as perguntas norteadas no início dessa análise, visando compreender melhor o desempenho acadêmico de alunos em relação ao cotidiano, alimentação e conviencia familiar.
