## Dicionário de dados

| Campo original | Campo padronizado | Tipo | Categoria |
|---|---|---|---|
| `gender` | `gender` | STRING | Dimensão |
| `race/ethnicity` | `race_ethnicity` | STRING | Dimensão |
| `parental level of education` | `parental_level_of_education` | STRING | Dimensão |
| `lunch` | `lunch` | STRING | Dimensão |
| `test preparation course` | `test_preparation_course` | STRING | Dimensão |
| `math score` | `math_score` | INT | Medida |
| `reading score` | `reading_score` | INT | Medida |
| `writing score` | `writing_score` | INT | Medida |
| — | `average_score` | DOUBLE | Medida derivada |


```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SCHEMA}")
rows = [
("gender","gender","STRING","Dimensão","female; male"),
("race/ethnicity","race_ethnicity","STRING","Dimensão","group A; group B; group C; group D; group E"),
("parental level of education","parental_level_of_education","STRING","Dimensão","some high school; high school; some college; associate's degree; bachelor's degree; master's degree"),
("lunch","lunch","STRING","Dimensão","standard; free/reduced"),
("test preparation course","test_preparation_course","STRING","Dimensão","none; completed"),
("math score","math_score","INT","Medida","0-100"),
("reading score","reading_score","INT","Medida","0-100"),
("writing score","writing_score","INT","Medida","0-100"),
("—","average_score","DOUBLE","Medida derivada","0-100")]
df_dictionary=spark.createDataFrame(rows,["source_column","standard_column","data_type","category","expected_domain"])
df_dictionary.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable(DATA_DICTIONARY_TABLE)
display(df_dictionary)
```
